```python
import matplotlib.pyplot as plt
import pandas as pd
import streamlit as st

# Sample data
data = {
    "label": ["high", "medium", "low", "high", "low", "medium", "high"]
}
df = pd.DataFrame(data)

counts = df["label"].value_counts()

# Colors for each category
colors = ["#ff4d4d", "#ffd11a", "#4dff88"]  # red, yellow, green

# Plot
fig, ax = plt.subplots()

ax.pie(
    counts.values,
    labels=counts.index,
    autopct='%1.1f%%',
    colors=colors,
    startangle=90,
    shadow=True,
    explode=(0.05, 0.05, 0.05)  # pop-out effect
)

ax.set_title("📊 Email Priority Distribution")

# Streamlit display
st.pyplot(fig)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample data
data = {
    "label": ["high", "medium", "low", "high", "low", "medium", "high"]
}
df = pd.DataFrame(data)

# Count values
counts = df["label"].value_counts()

# Plot
plt.bar(counts.index, counts.values)
plt.xlabel("Priority")
plt.ylabel("Count")
plt.title("Email Priority Distribution")
plt.show()
```
